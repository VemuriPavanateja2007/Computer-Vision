```python
img = cv2.imread('lena.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
zoomed = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_NEAREST)
```

```python
shrunk = cv2.resize(img, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_NEAREST)
```

```python
print("Original shape:", img.shape)
print("Zoomed shape:", zoomed.shape)
print("Shrunk shape:", shrunk.shape)
plt.figure(figsize=(12,4))
```

```python
plt.figure(figsize=(15,10))
plt.subplot(1,3,1), plt.imshow(img, cmap='gray'), plt.title("Original Image")
plt.subplot(1,3,2), plt.imshow(zoomed, cmap='gray'), plt.title("Zoomed Image")
plt.subplot(1,3,3) , plt.imshow(shrunk, cmap='gray'),plt.title("Shrunk Image")
plt.show()
```
