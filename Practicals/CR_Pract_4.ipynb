```python
# Load two images of the same size (grayscale)
img1 = cv2.imread('lion.jfif', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('tiger.jpg', cv2.IMREAD_GRAYSCALE)   # using same image for demo                                                  # (you can use any two images of same size
```

```python
# Make sure both images have the same size
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))
```

```python
# 1. Addition
added = cv2.add(img1, img2)
```

```python
# 2. Subtraction
subtracted = cv2.subtract(img1, img2)
```

```python
# 3. Multiplication (increase brightness by 1.5 times)
multiplied = cv2.multiply(img1, 1.5)
multiplied = np.clip(multiplied, 0, 255).astype(np.uint8)
```

```python
# 4. Blending (70% of img1 + 30% of img2)
blended = cv2.addWeighted(img1, 0.7, img2, 0.3, 0)
```

```python
# Print shapes
print("Image 1 shape:", img1.shape)
print("Image 2 shape:", img2.shape)
print("Added shape:", added.shape)
print("Subtracted shape:", subtracted.shape)
print("Multiplied shape:", multiplied.shape)
print("Blended shape:", blended.shape)
```

```python
# Display results
plt.figure(figsize=(15, 7))
plt.subplot(2, 3, 1), plt.imshow(img1, cmap='gray'), plt.title('Image 1')
plt.subplot(2, 3, 2), plt.imshow(img2, cmap='gray'), plt.title('Image 2')
plt.subplot(2, 3, 3), plt.imshow(added, cmap='gray'), plt.title('Added')
plt.subplot(2, 3, 4), plt.imshow(subtracted, cmap='gray'), plt.title('Subtracted')
plt.subplot(2, 3, 5), plt.imshow(multiplied, cmap='gray'), plt.title('Multiplied (×1.5)')
plt.subplot(2, 3, 6), plt.imshow(blended, cmap='gray'), plt.title('Blended (70% + 30%)')
plt.tight_layout()
plt.show()
```
