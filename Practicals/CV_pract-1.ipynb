```python
img = cv2.imread('/content/lena.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
levels =4
step = 256 // levels
reduced = (img // step) * step
reduced = reduced.astype(np.uint8)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1), plt.imshow(img, cmap='gray'), plt.title('Original (256 levels)')
plt.subplot(1,2,2), plt.imshow(reduced, cmap='gray'), plt.title('Reduced (4 levels)')
plt.show()
```

```python
levels_list = [256, 128, 32, 8, 4, 2]
images = []
for levels in levels_list:
    if levels == 256:
        images.append(img)              # Original image, no reduction
    else:
        step = 256 // levels
        reduced = (img // step) * step
        reduced = reduced.astype(np.uint8)
        images.append(reduced)

```

```python
plt.figure(figsize=(15, 8))                     # Set the overall figure size

for i in range(len(images)):
    plt.subplot(2, 3, i+1)                       # Create a 2x3 grid of subplots
    plt.imshow(images[i], cmap='gray')           # Show image in grayscale colormap
    plt.title(f"{levels_list[i]} Levels")        # Title showing number of levels
    plt.axis('off')                              # Hide axis ticks/labels

plt.tight_layout()   # Adjust spacing so titles/images don't overlap
plt.show()            # Render the final output

```
