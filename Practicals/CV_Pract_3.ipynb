```python
img = cv2.imread('lena.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
# Zoom in 2x using Bilinear Interpolation
zoomed = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_LINEAR)
```

```python
# Shrink to half size using Bilinear Interpolation
shrunk = cv2.resize(img, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)
```

```python
# Print image dimensions to confirm size change
print("Original shape:", img.shape)
print("Zoomed shape:", zoomed.shape)
print("Shrunk shape:", shrunk.shape)


```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1), plt.imshow(img, cmap='gray'), plt.title('Original')
plt.subplot(1,3,2), plt.imshow(zoomed, cmap='gray'), plt.title('Zoomed 2x (Bilinear)')
plt.subplot(1,3,3), plt.imshow(shrunk, cmap='gray'), plt.title('Shrunk 0.5x (Bilinear)')
plt.show()




```
