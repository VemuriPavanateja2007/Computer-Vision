```python

# Load grayscale image (keep it as uint8)
img = cv2.imread('lena.jpg', cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Error: Image not found. Check the file name/path.")
else:
    # Step 1: Apply Laplacian (on uint8 image)
    laplacian = cv2.Laplacian(img, cv2.CV_64F, ksize=3)
```

```python
    # Step 2: Sharpen by adding Laplacian back to original
    sharpened = img.astype(np.float64) + laplacian
    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)

    # For display
    laplacian_disp = np.uint8(np.clip(np.absolute(laplacian), 0, 255))

```

```python
# Display results
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1), plt.imshow(img, cmap='gray'), plt.title('Original')
plt.subplot(1, 3, 2), plt.imshow(laplacian_disp, cmap='gray'), plt.title('Laplacian (Edges)')
plt.subplot(1, 3, 3), plt.imshow(sharpened, cmap='gray'), plt.title('Sharpened (Original + Laplacian)')
plt.tight_layout()
plt.show()
```
