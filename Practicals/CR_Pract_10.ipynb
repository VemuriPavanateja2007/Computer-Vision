```python
# Load grayscale image
img = cv2.imread('Data For CV/lena.jpg', cv2.IMREAD_GRAYSCALE)

# ---------------- Sobel edge detection ----------------
sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
sobel_mag = cv2.magnitude(sobel_x, sobel_y)
sobel_disp = cv2.normalize(sobel_mag, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

# Threshold the Sobel magnitude to get a binary edge map
_, sobel_edges = cv2.threshold(sobel_disp, 60, 255, cv2.THRESH_BINARY)

# ---------------- Canny edge detection ----------------
canny_edges = cv2.Canny(img, 100, 200)   # (image, lowThreshold, highThreshold)

# Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1), plt.imshow(img, cmap='gray'), plt.title('Original')
plt.subplot(1, 3, 2), plt.imshow(sobel_edges, cmap='gray'), plt.title('Sobel Edges (thresholded)')
plt.subplot(1, 3, 3), plt.imshow(canny_edges, cmap='gray'), plt.title('Canny Edges')

plt.tight_layout()
plt.show()
```
